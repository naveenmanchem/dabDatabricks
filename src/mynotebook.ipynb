This notebook is managed and deployed via Databricks Asset Bundles (DABs) and GitHub Actions.

```python
# Create dummy data to test the pipeline execution
data = [
    ("Alice", 34, "Data Engineer"),
    ("Bob", 45, "Data Scientist"),
    ("Charlie", 29, "Analytics Engineer")
]
columns = ["Name", "Age", "Role"]

df = spark.createDataFrame(data, schema=columns)
display(df)
```

```python
# Perform a basic PySpark filtering transformation
from pyspark.sql.functions import col

filtered_df = df.filter(col("Age") > 30)
display(filtered_df)
```
